# 04 · DuckDB UI  ⏱ 10 min

Desde a 1.2.1 o DuckDB tem uma **interface web local** própria (extensão `ui`), no estilo notebook:

- células SQL com autocompletar e resultados em tabela;
- explorador do catálogo (bancos anexados, schemas, tabelas, colunas);
- **diagnóstico de colunas**: histograma, nulos, valores distintos, com um clique;
- tudo roda **na sua máquina**, em `http://localhost:4213`. Os dados não saem do computador.

> Na primeira vez a extensão `ui` é baixada, então precisa de internet.

## Jeito 1: pelo terminal (recomendado para a demo)

```bash
duckdb -ui curso.duckdb
```

O navegador abre sozinho. Feche com `Ctrl+C` no terminal.

## Jeito 2: a partir do Python
Útil quando os dados já estão numa conexão do notebook.

````python
import duckdb
con = duckdb.connect("curso.duckdb")

try:
    con.sql("CALL start_ui()").show()
except Exception as e:
    print("Não foi possível iniciar a UI:", str(e).splitlines()[0][:150])
````

Enquanto esta conexão estiver aberta, a UI fica disponível. Para encerrar:

````python
try:
    con.sql("CALL stop_ui_server()").show()
except Exception as e:
    print(str(e).splitlines()[0][:150])
````

## Roteiro da demonstração (5 minutos)

1. Abra `curso.duckdb` na UI e mostre o **catálogo** à esquerda (`clientes`, `produtos`, `vendas`).
2. Clique em `vendas` e mostre o **resumo das colunas**: distribuição de `canal`, faixa de `data_venda`, nulos.
3. Crie uma célula e rode:
   ```sql
   SELECT date_trunc('month', data_venda) AS mes, sum(quantidade * preco_unitario) AS receita
   FROM vendas GROUP BY ALL ORDER BY mes;
   ```
4. Clique na coluna `receita` do resultado para ver o histograma.
5. Mostre que dá para ler arquivo direto: `FROM 'dados/logs/*.jsonl' LIMIT 10;`
6. Mostre uma variável com a sintaxe nova da 2.0:
   ```sql
   SET VARIABLE canal_alvo = 'app';
   SELECT count(*) FROM vendas WHERE canal = $canal_alvo;
   ```
